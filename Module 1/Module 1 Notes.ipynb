# Generalization
We do need labeled data for testing and for model building, it SHOULD NOT BE THE SAME DATA
* Yeah, if it can predict all the training variables with no error, that is not impressive
    * It is probably just overfit for the data, we want to know how well can they predict NEW data?
* Get a variety of test points and see what performs better
* making new predictions is called generalization
* randomly sort the labeled examples into buckets is good practice, use half for test, use half for fitting
* all things being equal, the SIMPLER is the BETTER one (Occcam's Razor)

# Machine Learning Framing
* logical functions - outputs are deterministic, testing is logical and tests extremes
* statistical functions - outputs are predictions, testing is statistical and typically checks average case
    * ML depends on labeled data
    * Machine learning lives more here

#### Framing

# Example 1. Lung Cancer Screening
Practical example of machine learning
* Input: CT scans
* Output: probability of cancer, image regions and probabilities (might be more helpful)
* labeled data: not ground truth, but we have historical data -> doctor annotations
    * could get agreement rates for doctors with differing opinions to get a maximum predictability
    * multiple experts and agreement
* Test/Train Split: no patient with scans in both (in data set, person could have multiple scans in there, dont want it to predict too well thats misleading)
    * could be a patient split instead
* How will it be used: By doctors, by patients, by insurance, does training pop match usage pop?
    * trained in america, used in asia for example
* Evaluate predictions
    * Classification accuracy: correct/total
    * False positives vs. false negatives -> depending on who is using this it could mean different things
    * Calibration is what does a prediciton of 0.9, is that 90% change of cancer? or does it mean something else
        * gap between predicted and the baseline predictor
* Baseline predictor
    * always predict "no cancer", how well does this model do? (if its right 90% of the time with 900 no cancer and 100 cancer)



# Example 2: Sports Outcomes
* Input: team 1 and team 2
* Outputs: win/loss or point difference
* Labeled Data: previous games and results from the current season
* Test/Train Split: use past games to predict future (rolling predictions)
    * first 25 days is training data, next 5 are testing
* How will it be used: gambling, for the team
* Evaluate: Accuracy/MSE
* Baseline predictor: team win/loss records 

# Example 3. Energy Usage
Load on a powergrid that changes throughout the day
* Input: date, time, weather (15 min intervals)
* Output: energy usage
* Labeled data: previous observations from years past
* Train/Test split: rolling predictions, use past year, make prediciton about the next day, then use all that to predict the next
* How will it be used: Coordinate mixture of power sources, can only use solar during certain amounts of time
