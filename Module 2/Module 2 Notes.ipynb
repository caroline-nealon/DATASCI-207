## Gradient Descent - Intuition

Trying to climb a mountain in the dark -> taking steps and feeling around to see where it is the steepest possible

### One variable loss
Find w that minimizes J
* instead of evaluating at every value of W, start somewhere (anywhere)
* how do we mathematically decide which direction we want to go?
* positive gradient -> should be smaller (<w)
* negative gradient -> should be larger (>w)

Gradient is given by d/dw (J(w)) (that is the partial derivative)
corresponding update: 

w := w - alpha* dJ(w)/dw

alpha = learning rate (size of step to take between them)
* If it is too large, J diverges
* If its too small it takes sooooo long


## Gradient Descent: Derivation

$\frac{\partial J(w)}{\partial w}$

Gradeint for w1

$J(w_0,w_1) = \frac{1}{m}\sum{(w_0 + w_1*x_i - y_i)^2}$

-> 
$\frac{\partial J}{\partial w1} = 2(\hat{y} - y)x$

For w0 it turns out to
$\frac{\partial J}{\partial w0} = 2(\hat{y} - y)$

## Gradient Descent: In Practice
Stochastic Gradient Descent

first, estimate the gradient from all data
* Inefficient for large data sets
* stochastic estimates gradient from 1 example

Mini batchc SGD: estimate gradient from small batches of examples
* this is what is used in practice
* batch size is another parameter (hyper parameter)
* people refer to this as SGD

#### Epochs?
A single pass through the data is an epoch
* every data point gets a vote

How many epochs do you need?
* Confused on this?
* Answers the question of how long do you train for

Not all loss functions are differentiable
* absolute loss is not differentiable because it has the point!

Local Minima

What about plateaus?
